# Código-fonte RPA (Pyhton)

````python
import requests
from bs4 import BeautifulSoup

URL_PAGINA = "https://digitalinnovationone.github.io/dio-lab-assistente-investimentos-rpa-n8n/"

# Para usar a pagina do seu fork, habilite o GitHub Pages (Settings > Pages > /docs)
# e descomente a linha abaixo:
# URL_PAGINA = "https://evfernandes-dev.github.io/dio-lab-assistente-investimentos-rpa-n8n/"

# 1) Baixar a pagina
html = requests.get(URL_PAGINA, timeout=30).text

# 2) Ler a tabela de clientes
soup = BeautifulSoup(html, "html.parser")
linhas = soup.select("#clientes tbody tr")

clientes = []

for linha in linhas:
    colunas = linha.find_all("td")
    cliente = {
        "nome": colunas[0].get_text(strip=True),
        "email": colunas[1].get_text(strip=True),
        "saldo": colunas[2].get_text(strip=True),
        "perfil": colunas[3].get_text(strip=True),
    }
    clientes.append(cliente)

print(f"Clientes encontrados: {len(clientes)}")
for cliente in clientes:
    print(cliente)

# 3) Enviar para o Webhook do N8N
# Local: http://127.0.0.1:5678/webhook/clientes
# No N8N Cloud / Google Colab: use a URL publica do seu webhook
N8N_WEBHOOK = "http://127.0.0.1:5678/webhook/clientes"

resposta = requests.post(N8N_WEBHOOK, json={"clientes": clientes}, timeout=180)
print("Status:", resposta.status_code)
print("Resposta:", resposta.text)
````
